```python
# np.isnan(array_name) --> used to identify the errors related to calculation like (0/0, etc.) or when a value is missing from the dataset.
import numpy as np

arr1 = np.array([1,2,3,np.nan,4,5,np.nan,1,0,7,np.nan])
error = np.isnan(arr1)
print(error)
```

```python
# np.nan_to_num(array_name, nan=value) --> is used to fill the nan spaces with a value

arr1 = np.array([1,2,3,np.nan,4,5,np.nan,1,0,7,np.nan])
error = np.nan_to_num(arr1, nan=3) #and if we do not pass any value in the bracket then the default value i.e. '0' is printed as output
print(error)
```

```python
# np.isinf() --> to find the values which are infinite such as 10^10000, 1/0, etc.

arr2 = np.array([1,2,np.inf,4,-np.inf])
cleaned = np.isinf(arr2)
print(cleaned)

# how to replace infinite values with finite values

replaced = np.nan_to_num(arr2, posinf=1000, neginf=-1000)
print(replaced)
```
